## **Python interpreter**

The **Python interpreter** is a computer program that executes code written in the Python programming language. An **interactive interpreter** is a program that allows the user to execute _one line of code at a time_.

**Code** is what we call the instructions to be executed by a computer. A **line** of code is any line of text in code.

The interactive interpreter displays a prompt (`>>>`) which indicates that the interpreter is ready to accept code. The user types a line of Python code and presses the enter key to execute the code. The example program below calculates a salary based on a given hourly wage, the number of hours worked per week, and the number of weeks per year. ## Executing a Python program
 The Python interactive interpreter is useful for simple operations or programs consisting of only a few lines. However, entering code line-by-line quickly becomes impractical for any program spanning more than a few lines.

Instead, a programmer can write Python code in a file and provide that file to the interpreter. The interpreter begins by executing the first line of code at the top of the file and continues until the interpreter reaches the end. 

In this class, we'll be learning Python through interactive Python notebooks via Google Colab. Refer to [Section 1.0](https://cse20-textbook.readthedocs.io/en/latest/1-0/) if you need a refresher on Colab!


### Statements
A **statement** is a program instruction. A program mostly consists of statements, and each statement usually appears on its own line.

```python
# example statement
print("Welcome to CSE 20!! I hope you enjoy this course :)")
```

**Variables** are named references to values _stored_ by the interpreter.

In the example below, the names `wage`, `hours`, `weeks`, and `salary`are variables.


```python
wage = 25
hours = 40
weeks = 48
```

An **expression** is a piece of code the interpreter evaluates to produce a value. In this example, the code `wage * hours * weeks` is an expression that computes a number. (The symbol `*` is an operation used for multiplication; we'll learn more about operations later!)

The `print()` function displays variables or expression values.

```python
salary = wage * hours * weeks
print(salary)
```

Characters such as `#` denote **comments**, which are optional but can be used to explain portions of code to a human reader.

```python
# this is a comment!
print("The line above is a comment!")
```
