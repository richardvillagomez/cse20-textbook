## Variables and assignments
In a program, a **variable** is a named item that holds a value.

An assignment statement assigns a variable with a value, such as `x = 5`. That statement means `x` is assigned the value 5, and `x` keeps that value during subsequent statements, until `x` is assigned again.

An assignment statement's left side _must be a variable_. The right side can be a value, or an expression that evaluates to a value.

Say we _define_ the following three variables: 

```python
mascot_name = "Sammy"   
mascot_age = 40        
mascot_height = 70.5
```

Now that we have defined our variables, we can refer to them using their **variable names**. Below, we are simply printing out the values of the variables.

```python
print(mascot_name)
```

```python
print(mascot_age)
```

```python
print(mascot_height)
```

In addition, we can _update_ the values of our variables. Say our mascot grows a year older and we want to increment his `age` value by 1. We can do so by referencing its current value, adding 1 to it, and assigning that value to `mascot_age`:

```python
mascot_age = mascot_age + 1
print(mascot_age)
```

One important note: in programming, `=` is used to assign a value to a variable, rather than the way it is used in math equations. Notice that the following line of code errors: 


```python
2 + 2 = 4
```
