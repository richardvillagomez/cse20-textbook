## Computer programs
Computers and programs are *everywhere*! Programs affect many aspects of daily life and society as a whole. We depend on programs for communication, shopping, finances, healthcare, education, entertainment... hopefully you get the gist. Learning how to program computers opens the door to countless opportunities for building a better world.

A computer program consists of **instructions**. Basic instruction types are:
- **Input**: A program receives data .
- **Process**: A program performs computations on that data.
- **Output**: A program puts that data somewhere.

Python consists of **built-in functions**, which allow you to perform a specific task when called. In this section, we'll focus on two functions: `print()` and `input()`.

---

## Basic output
The `print()` function _displays output_ to the user. Output is the information or result produced by a program. Try running the following cell: 

````python
print("Welcome to CSE 20! I hope you have a great time :)")
````

By calling `print()` on a message enclosed in _quotation marks_ (known as **strings** in Python), we are able to display our message to the user.

Multiple values, separated by commas, can be printed in the same statement. By default, each value is separated by a space character in the output.

````python
print("This textbook was written in Fall 2026.", "I hope you find it useful!")
````

By default, the `print()` function adds a _newline_ character at the end of the output. A newline character tells the display to move to the next line.

````python
print("Today is Tuesday.")
print("I like chocolate!")
````

## Basic input
Computer programs often receive input from the user. **Input** is what a user enters into a program. An input statement, `variable = input("prompt")`, has three parts:
1. A variable refers to a value stored in memory. In the statement above, variable can be replaced with any name the programmer chooses.
2. The `input()` function reads one line of input from the user. The input is stored in the computer's memory and can be accessed later using the variable.
3. A prompt is a short message that indicates the program is waiting for input. In the statement above, "prompt" can be omitted or replaced with any message.

Try running the following line of code, and enter your own input when prompted:


````python
# run this cell, and enter your favorite restaurant as input!
favorite_restaurant = input("Enter the name of your favorite restaurant: ")
print(favorite_restaurant, "is a good choice!")
````

You probably noticed the **comment** in the above cell, which consists of `#` followed by the comment itself. Comments are _optional_ and do not change the functionality of your code. But it's good practice to use them to describe what your code does!
