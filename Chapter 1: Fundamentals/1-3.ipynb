Python displays text using the built-in function `print()`, with the text in quotes within the parentheses. 

In Python, text enclosed in quotes is known as a **string**. Text in strings may have letters, numbers, spaces, or symbols like. Below are a few examples of printed strings:


````python
print("Hello")
````

````python
print("Email sent!")
````

````python
print("jbask@ucsc.edu")
````

## Print and the newline character
Output can be moved to the next line using the newline character `"\n"`. An output newline is the equivalent of pressing the Enter key on a keyboard to get to the next line. For example, `print("1\n2\n3")` prints `"1"` on the first line, `"2"` on the second line, and `"3"` on the third line, as shown:



````python
print("1\n2\n3")
````

`print()` always outputs an invisible newline character at the end of the statement. For example: 

````python
print("Hello\nthere.")
print()
print("Nice to")
print("meet you.")
````

## Output variables and multiple values
The value of a variable can be printed out using `print(variable_name)` (_without quotes_ around the variable name). 

A single print statement may include multiple strings or variables, **separated by commas**. In the output, a single space is automatically added between each string and variable expression.

Take a look at the code below, and try to understand its output.

````python
wage = 400
print("Your daily wage is $", wage)

weekly_wage = 2000
print("$", wage, "per day is $", weekly_wage, "per week")
````

## Keeping output on the same line
Each use of `print()` outputs on a new line. However, sometimes a programmer may want to keep output on the same line. Adding `end=" "` inside of `print()` keeps the output of the next print on the same line, separated by a single space character. Any character, or multiple characters, can be used instead of a space character.

````python
# Including end=" " keeps output on same line
print("Hello there.", end=" ")
print("My name is...", end=" ")
print("Carla?")
````

## Escape Sequences
An escape sequence is a string that has a special meaning, like the newline character `\n`, that always starts with a backslash `\`. Other escape sequences exist, such as `\t` to insert a tab, or `\\`to print an actual backslash character.

Any space, tab, or newline is called **whitespace**.

````python
# notice the escape sequences in this code!
print("Name\tJob")
print("------------------")
print("Ann\tDeveloper\nJoe\tInfluencer")
````

## Basic input
Many useful programs allow a user to enter values, such as a name or a number. 

The `input()` function is used to read input from a user. It causes the program to wait until the user has entered text and pushed the return key. 

Take the following example: 

````python
print("Enter name of best friend:", end=" ")
best_friend = input()
print("My best friend is", best_friend)
````

In this example, the user inputted `Sijia`, which is then stored in the `best_friend` variable and printed.

## Converting input types
The string "123" (with quotes) is fundamentally different from the integer 123 (without quotes). The "123" string is a sequence of the characters "1", "2", and "3" arranged in a certain order, whereas 123 represents the integer value one-hundred twenty-three.

Strings and integers two examples of data types. **Data types** in Python define the type of value stored in a variable and determine the operations that can be performed on that data. For example, integers can be divided, but strings can't (what sense would `"Hello" / 3` make?). Data types are discussed in detail later on.

Reading from input is always stored as a string. However, often a programmer wants to read in an integer, and then use that number in a calculation. If a string contains only numbers, like "123", then the `int()` function can be used to convert that string to the integer 123.
