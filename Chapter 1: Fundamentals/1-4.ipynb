## Syntax Errors
No matter your level of experience, it's totally normal to make mistakes in programming. (Mistakes often make a great opportunity for learning!) In programming, these mistakes are called **errors**. 

> "Where there is code, there will be errors." - Unknown

A **syntax error** violates a programming language's rules on how symbols can be combined to create a program. 

The interpreter will generate a message when encountering a syntax error. The **error message** will report the line number of the error, so you can know where the occurs.

Below, the message "invalid syntax" is not very precise, but is the best information that the interpreter provides. As you get more practice, you will become more familiar with common errors and how to fix them!

```python
current_salary = 45000
new_salary = 48000

print("Current salary is", end=" ")
print(current_salary) 

print("New salary is", end=" ")
print(new_salary) 

print(current_salary) print(new_salary)
```

Note that syntax errors are found **before** the program is run by the interpreter. In the example above, none of the print statements prior to the error are displayed.

## Good coding practice
It's normal for programmers of all levels of experience to make syntax errors. With that, it's generally good practice is to execute your code frequently, writing perhaps a few lines of code at a time, and checking it for errors. Still, remember that it's totally normal to make mistakes!

## Runtime errors
 The Python interpreter is able to detect syntax errors when the program is initially loaded, prior to actually executing any of the statements in the code. However, just because the program loads and executes does not mean the program is fully correct. Specifically, the program might have a **runtime error**, in which the syntax is correct but it attempts an impossible operation, such as dividing by zero or multiplying strings together (what sense would `"Hello" * "ABC"` make?). A runtime error halts the execution of the program, or "crashes" it.

Consider the below program, which begins executing, prints the salary, and then waits for the user to enter an integer value. The `int()` statement expects a number to be entered, but gets the text `"Henry"` instead. 

```python
print("Salary is", end=" ")
print(20 * 40 * 50)

print("Enter integer: ", end=" ")
user_num = int(input())
print(user_num)
```

Runtime errors are categorized into _types_ that describe the sort of error that has occurred. Above, a ValueError occurred, indicating that the wrong sort of value was passed into the `int()` function. Other examples include a NameError and a TypeError, both described in the table below.

| Error type       | Description                                                              |
|------------------|--------------------------------------------------------------------------|
| SyntaxError      | The program contains invalid code that cannot be properly understood.    |
| IndentationError | The lines of the program are not properly indented.                      |
| ValueError       | An invalid value is used, which can occur if giving letters to `int()`.  |
| NameError        | The program tries to use a variable that does not exist.                 |
| TypeError        | An operation uses incorrect or incompatible types.                       |


You are not expected to immediately memorize all types of errors. The more errors you run into, the more familiar you will get!

## Logic Errors
Some errors may be subtle enough to silently misbehave, instead of crashing the program. Take the following example: 

```python
current_salary = 10000

raise_percentage = 3  # logic error gives a 300% raise instead of 3%
new_salary = current_salary + (current_salary * raise_percentage)
print("New salary:", new_salary)
```

The programmer clearly made an error, but there are no issues with the syntax of the code. The interpreter will not generate an error message, but the program's output is not what the programmer expects: the new computed salary is too high. This is an example of a **logic error**, as the program is _logically flawed_. These are also known as bugs.

These mistakes can be very easy to miss. Paying careful attention and running code after writing just a few lines can go a long way in preventing runtime errors!
