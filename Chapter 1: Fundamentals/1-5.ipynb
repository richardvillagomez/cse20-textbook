## Variables and assignments
In a program, a **variable** is a named item that holds a value.

An assignment statement assigns a variable with a value, such as `x = 5`. That statement means `x` is assigned the value 5, and `x` keeps that value during subsequent statements, until `x` is assigned again.

An assignment statement's left side _must be a variable_. The right side can be a value, or an expression that evaluates to a value.

Say we _define_ the following three variables: 

```python
mascot_name = "Sammy"   
mascot_age = 40        
mascot_height = 70.5
```

Now that we have defined our variables, we can refer to them using their **variable names**. Below, we are simply printing out the values of the variables.

```python
print(mascot_name)
```

```python
print(mascot_age)
```

```python
print(mascot_height)
```

In addition, we can _update_ the values of our variables. Say our mascot grows a year older and we want to increment his `age` value by 1. We can do so by referencing its current value, adding 1 to it, and assigning that value to `mascot_age`:

```python
mascot_age = mascot_age + 1
print(mascot_age)
```

One important note: in programming, `=` is used to assign a value to a variable, rather than the way it is used in math equations. Notice that the following line of code errors: 


```python
2 + 2 = 4
```

## Identifiers

As a programmer, you will have to choose names for various items, such as variables (and also functions, described later). For example, `x = 5` uses the name "x" to refer to the value 5. An **identifier** is made up of a sequence of letters (a-z, A-Z), underscores (_), and digits (0–9), and must start with a letter or an underscore.


Python is case sensitive, meaning uppercase and lowercase letters differ. Consider the following two variables and their identifiers: 

```python
bird_sound = "tweet."
Bird_sound = "TWEET!"
```

```python
print(bird_sound)
```

```python
print(Bird_sound)
```

Even though they are spelled the same, the difference in capitalization means they are different identifiers. 

Note the following examples of _invalid_ identifiers, and the reasoning behind why they are invalid: 

```python
# cannot start with number!
3d = "three dimensional!"
```

```python
# cannot include a space!
hi there = "hi there!"
```

```python
# cannot use symbols other than letters, digits, underscores
dog$ = "labrador, golden retriever!"
```



Names that start and end with double underscores (for example, `__init__`) are allowed but should be avoided because Python has special usages for double underscore names, explained elsewhere. A good variable name should describe the purpose of the variable, such as `height` or `age`, rather than just `h` or `A`.

Certain words like `and` or `True` cannot be used as names. _Reserved words_, or keywords, are words that are built into Python and cannot be used as a programmer-defined variable. You can find a list of reserved words in the [Python documentation](https://docs.python.org/3/reference/lexical_analysis.html#keywords).

## Style
Standard practice when naming variables in Python is to use _all lowercase letters_ and to place _underscores between words_. Having a consistent variable naming scheme helps ensure that other programmers can your code.

Here are a few other good practices: 
- When dealing with scientific or engineering names is to append the unit of measure. For example, instead of `temperature`, use `temperature_celsius`. 
- Abbreviations should only be used if widely understandable, as in `tv_model` or `ios_app. `
- Very long variable names, such as `average_age_of_a_ucsc_undergraduate_student`, can make subsequent statements too long and hard to read. 

As a programmer, it is your job to find a balance between descriptive names and succinct names. The [Python Style Guide](https://peps.python.org/pep-0008/#descriptive-naming-styles) is a great resource for achieving this!
